با توجه به رابطه زیر احتمال وجود یال میان دو گره راه محاسبه میکنیم. سپس هر دو گره را دو به دو بررسی کرده و با احتمالی که بدست آوردیم میان آن دو گره یالی برقرار میکنیم (با فرض اینکه شبکه بدون جهت است و از یک گره به خودش یالی نیست)

$$ <k> = p(N-1) $$

در آخر به کمک کتابخانه networkx شبکه بدست آمده را رسم میکنیم.

```python
import random
import numpy
import networkx
from matplotlib import pyplot

N = 500


def generate_network(k_avg):
    p = k_avg / (N - 1)
    network = [[0] * N for _ in range(N)]
    for i in range(N):
        for j in range(i + 1, N):
            if network[i][j] == 1:
                continue
            rand = random.random()
            if p >= rand:
                network[i][j] = 1
                network[j][i] = 1
    return network


def visualize_network(network):
    pyplot.figure(figsize=(30, 30))
    graph = networkx.from_numpy_array(numpy.matrix(network), create_using=networkx.Graph)
    networkx.draw(graph, )
    pyplot.show()
```

**a**

$ <k> = 0.8 $

```python
network = generate_network(0.8)
visualize_network(network)
```

**b**

$ <k> = 1 $

```python
network = generate_network(1)
visualize_network(network)
```

**c**

$ <k> = 8 $

```python
network = generate_network(8)
visualize_network(network)
```
